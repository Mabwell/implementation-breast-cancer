# Exploratory Data Analysis

In this notebook, we will perform exploratory data analysis (EDA) on the Breast Cancer Wisconsin dataset. We will visualize the data, check for missing values, and understand the distribution of features.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# Load the dataset
data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

# Display the first few rows of the dataset
df.head()
```

## Data Overview

Let's check the shape of the dataset and get some basic statistics.

```python
print(f'Shape of the dataset: {df.shape}')
print('Statistics:')
print(df.describe())
```

## Visualizations

Let's visualize the distribution of the target variable and some features.

```python
# Distribution of target variable
sns.countplot(x='target', data=df)
plt.title('Distribution of Target Variable')
plt.xlabel('Target (0: Benign, 1: Malignant)')
plt.ylabel('Count')
plt.show()
```

```python
# Pairplot of features
sns.pairplot(df, hue='target')
plt.title('Pairplot of Features')
plt.show()
```

## Conclusion

In this notebook, we performed exploratory data analysis on the Breast Cancer Wisconsin dataset. We visualized the distribution of the target variable and explored relationships between features.
