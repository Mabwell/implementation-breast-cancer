# Breast Cancer Wisconsin Dataset Exploration

In this notebook, we will perform exploratory data analysis (EDA) on the Breast Cancer Wisconsin dataset. We will visualize the data, check for missing values, and understand the distribution of features.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualisation style
sns.set(style='whitegrid')
```

```python
# Load the dataset
data_path = '../data/raw/breast_cancer_data.csv'  # Update with the correct path
df = pd.read_csv(data_path)

# Display the first few rows of the dataset
df.head()
```

```python
# Check for missing values
missing_values = df.isnull().sum()
missing_values[missing_values > 0]
```

```python
# Visualize the distribution of the target variable
plt.figure(figsize=(8, 6))
sns.countplot(x='diagnosis', data=df)
plt.title('Distribution of Diagnosis')
plt.xlabel('Diagnosis (M = Malignant, B = Benign)')
plt.ylabel('Count')
plt.show()
```

```python
# Visualize correlations between features
plt.figure(figsize=(12, 10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```
